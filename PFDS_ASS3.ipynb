```python
def sum_natural(n):
    # Base case
    if n == 0:
        return 0

    # Recursive case
    return n + sum_natural(n - 1)


# Test the program with different values
for n in [5, 10, 15]:
    total_books = sum_natural(n)
    print("Number of Shelves:", n)
    print("Total Number of Books =", total_books)
    print()
```
