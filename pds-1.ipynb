```python
def factorial(n):
    if n == 0 or n == 1:  # Base case
        return 1
    return n * factorial(n - 1)  # Recursive case


n = int(input("Enter the number of gift boxes: "))

if n < 0:
    print("Please enter a non-negative whole number.")
else:
    print("Total Possible Arrangements =", factorial(n))
```
